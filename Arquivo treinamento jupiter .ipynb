```python
vendas = [
    {"produto": "Caderno", "quantidade": 10, "preco_unitario": 15.00},
    {"produto": "Caneta", "quantidade": 25, "preco_unitario": 3.50},
    {"produto": "Mochila", "quantidade": 4, "preco_unitario": 120.00},
    {"produto": "Bolsa", "quantidade": 8, "preco_unitario": 18.00}
]

def calcular_total_venda(venda):
    return venda["quantidade"] * venda["preco_unitario"]

def calcular_faturamento(vendas):
    total = 0
    for venda in vendas:
        total += calcular_total_venda(venda)
    return total

def encontrar_maior_venda(vendas):
    if not vendas:
        return None
    maior_venda = vendas[0]
    for venda in vendas[1:]:
        if calcular_total_venda(venda) > calcular_total_venda(maior_venda):
            maior_venda = venda
    return maior_venda

print("RELATÓRIO DE VENDAS")
print("-" * 30)

for venda in vendas:
    total_produto = calcular_total_venda(venda)
    print(f'{venda["produto"]}: {venda["quantidade"]} unidades — R$ {total_produto:.2f}')

print("-" * 30)
print(f"Faturamento total: R$ {calcular_faturamento(vendas):.2f}")

maior_venda = encontrar_maior_venda(vendas)
if maior_venda is not None:
    print(f'Maior faturamento por produto: {maior_venda["produto"]}')
    print(f'Valor: R$ {calcular_total_venda(maior_venda):.2f}')
```
